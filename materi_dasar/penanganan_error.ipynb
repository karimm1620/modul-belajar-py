# Apa itu Error?
- Error adalah kesalahan yang terjadi saat program berjalan.
- Ketika error terjadi, program akan berhenti dan menampilkan error.

## 1. Syntax Error
- Kesalahan sintaks

```python
# Contoh syntax error
# Kurung tidak ditutup
print("Hello World" 
```

## 2. NameError
- Variabel tidak ditemukan

```python
# Contoh nama error
# Variable 'nama' belum di definisikan
print(nama)
```

## 3. TypeError
- Kesalahan tipe data

```python
# Contoh type error yang akan menyebabkan error:
# Tidak bisa menambah string dengan angka
print("5" + 5)  # Karena string tidak bisa ditambah angka/integer
```

## 4. ValueError
- Nilai tidak valid

```python
# Contoh value error yang akan menyebabkan error:
# Tidak bisa convert "abc" ke integer
angka = int("abc")
```

## 5. IndexError
- Index tidak valid / tidak ditemukan

```python
# Contoh index error yang akan menyebabkan error:
list_data = [1, 2, 3]
# Index 5 tidak ada
print(list_data[5])
```

## 6. KeyError
- Key tidak valid / tidak ditemukan

```python
# Contoh key error yang akan menyebabkan error:
data = {"nama": "Alice"}
# Key "umur" tidak ada
print(data["umur"])
```

## 7. ZeroDivisionError
- Pembagian dengan nol

```python
# Contoh zero division error yang akan menyebabkan error:
# Tidak bisa dibagi dengan 0
print(10 / 0)
```

# Penanganan Error
- Bagaimana untuk menangani error agar program kita tidak berhenti?
- Penanganan error memungkinkan program untuk menangani error dengan baik tanpa berhenti secara tiba-tiba

## 1. Try-Except
- try-except digunakan untuk menangani untuk error yang mungkin terjadi

```python
print("=== KALKULATOR SEDERHANA ===")

try:
    angka1 = int(input("Angka Pertama: "))
    angka2 = int(input("Angka Kedua: "))
    hasil = angka1 / angka2
    print("Hasil:", hasil)
except:
    print("Terjadi error dalam perhitungan!")
```

## 2. Menangani Error Spesifik
- Saat menangani error, mungkin kita ingin melakukan hal yang berbeda untuk jenis error yang berbeda

- Kita bisa menyebutkan tipe errornya setelah kata kunci except, sehingga jika terjadi error jenis tersebut, maka bagian except tersebut yang akan di eksekusi

- Kita bisa menambahkan lebih dari satu except

```python
try:
    a = int(input("Masukkan angka pertama: "))
    b = int(input("Masukkan angka kedua: "))
    hasil = a / b
    print("Hasil pembagian:", hasil)
except ValueError:
    print("Mohon masukkan angka yang valid!")
except ZeroDivisionError:
    print("Tidak bisa dibagi dengan nol!")
except:
    print("Terjadi kesalahan yang lain!")
```

## 3. Try-Except-Else
- try-except juga memiliki bagian else seperti pada if dan for

- Bagian else dijalankan jika tidak ada error pada try-except

- Jika terjadi error, bagian else tidak akan dijalankan

```python
try:
    angka = int(input("Masukkan angka: "))
except ValueError:
    print("Input tidak valid!")
else:
    print("Angka yang Anda masukkan:", angka)
    if angka > 0:
        print("Angka positif")
    elif angka < 0:
        print("Angka negatif")
    else:
        print("Angka nol")
```

## 4. Try Except Finally
- else di try-except hanya akan dijalankan ketika tidak ada error

- Bagaimana jika kita ingin melakukan sesuatu, baik itu jika terjadi error ataupun tidak?

- Kita bisa menggunakan finally

- finally selalu dijalankan, baik ada error atau tidak

```python
try:
    angka = int(input("Masukkan angka: "))
    print("Angka anda:", angka)
except ValueError:
    print("Input tidak valid!")
finally:
    print("Program selesai dijalankan")
```
