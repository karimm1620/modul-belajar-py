```python
# Tipe data di python
# Int => integer yaitu berisi angka bilangan bulat negatif dan positif
# contoh
umur = 25
tahun_lahir = 1970
my_saldo_gweh = -3000
nol = 0

# Bilangan besar (py bisa ngehandle bilangan yang besar)
# Contoh
populasi_sawit = 780000000
angka_acak_besar = 123456789012345678901234567890

print(type(umur))           # <class 'int'>
print(type(my_saldo_gweh))  # <class 'int'>
```

```python
# Float adalah tipe data dengan bilangan koma
# Bilangan dengan koma
tinggi = 170.5
berat = 45.7
pi = 3.14159
suhu = -10.5

# Notasi scientific
speed_of_light = 3e8        # 3 x 10^8 = 300000000
very_small = 1e-6           # 0.000001

print(type(tinggi))         # should print <class 'float'>
print(speed_of_light)       # should print 300000000.0
```

```python
# String => adalah tipe data berupa teks
# Contoh

# String single quotes
alamat = 'Jl. Bebulak, ds. marga mulya'
pesan = 'Welcome to bebulak'

# String double quotes
nama = "claude"
kota = "Bandung"

# String multi quotes (Multi-Line)
puisi = """
ini adalah puisi
Yang terdiri dari
Beberapa baris
"""

# String kosong
empty_string = ""
another_empty = ''

print(type(nama))

print(puisi)
```

```python
# Boolean => adalah tipe data yang bernilai 'True' atau 'False'

# Boolean values

is_kerja = False
is_nganggur = True

print(type(is_kerja))
```

```python
# Assignment dasar
nama = 'Karim'
umur = 18
tinggi = 170.5
is_kerja = False

# Multiple assignment
x = y = z = 0             # Semua variabel bernilai '0'
a, b, c = 1, 2, 3         # a=1, b=2, c=3
name, age = "Bob", 23     # name="Bob", age=23

# Mengubah nilai variabel
skor = 80     # nilai pertama
print(skor)   # 80
skor = 95     # mengubah nilai menjadi 95
print(skor)   # 95
```

# Konversi tipe data
- konversi tipe data adalah mengubah data dari satu tipe ke tipe lainnya.
- Terdapat beberapa function yang bisa kita gunakan untuk melakukan konversi dari satu tipe data ke tipe data lain:
  - int(), mengubah string atau data lain menjadi integer (bilangan bulat)
  - str(), mengubah integer atau data lain menjadi string (teks)
  - float(), mengubah string atau integer menjadi float (bilangan desimal)
  - bool(), mengubah data lain menjadi boolean (True/False)

```python
nama = input("Masukkan nama anda: ")
print("Hello", nama)

umur_teks = input("Masukan umur anda: ")
umur = int(umur_teks)
print("Umur anda:", umur)
print("Tipe datanya adalah:", type(umur))
```
